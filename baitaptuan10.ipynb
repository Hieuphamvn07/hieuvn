```python
def happy_number(n):
    seen = set()

    while n != 1 and n not in seen:
        seen.add(n)
        s = 0
        while n > 0:
            digit = n % 10
            s += digit ** 2
            n //= 10
        n = s

    return n == 1

try:
    a = int(input())

    if happy_number(a):
        print("Happy Number")
    else:
        print("Not Happy Number")

except ValueError:
    print("ValueError")
```

```python
import math

try:
    x = float(input())

    if x < 0:
        raise ValueError

    print(f"{math.sqrt(x):.2f}")

except ValueError:
    print("So am")
```

```python
try:
    N = int(input())

    arr = list(map(int, input().split()))

    if len(arr) != N:
        raise ValueError

    for x in arr:
        if x <= 0:
            raise ValueError

    print(*arr)

except ValueError:
    print("Mang khong hop le")
```
