```python
def Max_Of_True(a: int,b: int):
    if a > b:
        return a
    else:
        return b
```

```python
from typing import Tuple

def swap(a: int, b: int) -> Tuple[int, int]:
    return b, a
```

```python
def is_prime(n: int) -> bool:
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
def is_perfect(n: int) -> bool:
    if n <= 0:
        return False
    s = 0
    for i in range(1, n):
        if n % i == 0:
            s += i
    return s == n
```

```python
def find_first(lst: list[int], k: int) -> int:
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1
```

```python
def factorial(n: int) -> int:
    if n < 0:
        return None  # hoặc raise lỗi
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
```
