```python
def binary_search(arr, left, right, target):
    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
arr = list(map(int, input().split()))
target = int(input())

print(binary_search(arr, 0, len(arr) - 1, target))
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(n-1-i):
            if arr[i]>arr[j+1]:
                arr[i],arr[j+1]=arr[j+1],arr[i]
    return arr
```

```python
def most_frequent(arr):
    max_count = 0
    result = arr[0]

    for i in arr:
        count = arr.count(i)

        if count > max_count:
            max_count = count
            result = i

    return result, max_count


arr = list(map(int, input().split()))

num, times = most_frequent(arr)

print(f"{num} xuat hien nhieu nhat, {times} lan")
```

```python
def count_pairs(arr, x):
    count = 0

    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] + arr[j] == x:
                count += 1

    return count


arr = list(map(int, input().split()))
x = int(input())

print(count_pairs(arr, x))
```

```python
def flatten_list(arr):
    result = []

    for sublist in arr:
        for item in sublist:
            result.append(item)

    return result


arr = eval(input())

print(flatten_list(arr))
```
