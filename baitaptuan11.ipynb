```python
def reverse(path: str) -> int:
    f = open(path, "r")
    n = f.read().strip()
    f.close()

    return int(n[::-1])
```

```python

def canWinNim(path: str) -> bool:
    f = open(path, "r")
    n = int(f.read().strip())
    f.close()

    return n % 4 != 0
```

```python
def housesOfHogwarts(path: str):
    f = open(path, "r", encoding="utf-8")

    n = int(f.readline().strip())

    for i in range(n):
        print(f.readline().strip())

    f.close()
```

```python
def defAnagrams(path1: str, path2: str) -> bool:
    f1 = open(path1, "r", encoding="utf-8")
    s1 = f1.read()
    f1.close()

    f2 = open(path2, "r", encoding="utf-8")
    s2 = f2.read()
    f2.close()

    s1 = ''.join(c.lower() for c in s1 if c.isalnum())
    s2 = ''.join(c.lower() for c in s2 if c.isalnum())

    return sorted(s1) == sorted(s2)
```

```python
def count(path: str, character: str) -> int:
    f = open(path, "r", encoding="utf-8")
    sentence = f.read()
    f.close()

    return sentence.count(character)
```

```python
def countCharacter(path: str, letter: str) -> int:
   
    target_letter = letter.lower()
    count = 0
    
    try:
      
        with open(path, 'r', encoding='utf-8') as file:
            for line in file:
  
                count += line.lower().count(target_letter)
                
    except FileNotFoundError:
        print(f"Lỗi: Không tìm thấy tệp tin tại đường dẫn {path}")
        return 0
        
    return count
```
