```python
def merge_sorted_lists(list1, list2):
    result = []
    i, j = 0, 0
    
    # Duyệt qua cả hai danh sách bằng hai con trỏ
    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            result.append(list1[i])
            i += 1
        else:
            result.append(list2[j])
            j += 1
            
    # Nếu list1 còn phần tử, nốt vào kết quả
    while i < len(list1):
        result.append(list1[i])
        i += 1
        
    # Nếu list2 còn phần tử, nốt vào kết quả
    while j < len(list2):
        result.append(list2[j])
        j += 1
        
    return result
list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))
print(merge_sorted_lists(list1, list2))
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

n = int(input())
print(factorial(n))
```

```python
def find_number(lst, x):
    for num in set(lst):
        if lst.count(num) > x:
            return num
    return "INVALID"

lst = list(map(int, input().split()))
x = int(input())

print(find_number(lst, x))
```

Tính base^exp bằng đệ quy chia để trị

```python
def power(base, exp):
    if exp == 0:
        return 1

    half = power(base, exp // 2)

    if exp % 2 == 0:
        return half * half
    else:
        return half * half * base

base = int(input())
exp = int(input())

print(power(base, exp))
```
